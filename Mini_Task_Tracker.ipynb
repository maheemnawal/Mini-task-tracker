```python
import json
Filename="tasks.json"

#loading the json file
def load_json():
  try:
    with open(Filename, "r") as file:
      return json.load(file)
  except FileNotFoundError:
    return []

#saving the tasks in file
def save_tasks():
    with open(Filename, "w") as file:
        json.dump(tasks, file, indent=4)

#adding the tasks
def add_tasks():
  title=input("enter the task title:")
  if title == "":
      print("Task title cannot be empty.")
      return

  tasks.append({"title": title, "done": False})
  save_tasks()
  print("Task added!")

#viewing the tasks
def view_tasks():
  if not tasks:
        print("No tasks found.")
        return
  for number, task in enumerate(tasks, start=1):
        status = "Done" if task["done"] else "Pending"
        print(f"{number}. {task['title']} [{status}]")


#completing the tasks
def complete_task():
    view_tasks()

    if not tasks:
        return

    try:
        number = int(input("Task number: "))

        if number < 1 or number > len(tasks):
            print("Invalid task number.")
            return

        tasks[number - 1]["done"] = True
        save_tasks()
        print("Task completed.")

    except ValueError:
        print("Please enter a valid number.")

#deleting the tasks
def delete_task():
    view_tasks()

    if not tasks:
        return

    try:
        number = int(input("Task number to delete: "))

        if number < 1 or number > len(tasks):
            print("Invalid task number.")
            return

        removed_task = tasks.pop(number - 1)
        save_tasks()
        print(f"Deleted: {removed_task['title']}")

    except ValueError:
        print("Please enter a valid number.")

#loading previous tasks from the file
tasks = load_json()


#menu for the task tracker
while True:
    print("\n--- welcome to TASK TRACKER ---")
    print("1. Add  2. View  3. Complete  4. Delete  5. Exit")

    choice = input("Choice: ").strip()

    if choice == "1":
        add_tasks()
    elif choice == "2":
        view_tasks()
    elif choice == "3":
        complete_task()
    elif choice == "4":
        delete_task()
    elif choice == "5":
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Please select 1-5.")
```
